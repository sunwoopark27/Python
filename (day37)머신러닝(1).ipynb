## 머신러닝의 핵심

    - 학습 오차(Training Error)를 최소화하는 모델
    -
$MSE(traing) = (y-\hat{y})^2$

    -
$Expected MSE(test) = E[(y-\hat{y})^2 | x]$

### 주요 용어
- 특성(Feature)
    - 통계에선 독립변수라는 용어로 사용되며, 머신러닝에서는 특성(Feature)이라고 함
    - 설명변수, 위험인자라고도 함

- 레이블/라벨(Label)
    - 통계에선 종속변수라는 용어로 사용되며, 머신러닝에서는 레이블/라벨(Label), 클래스(Class)로 사용됨
    - 응답변수, 표적변수, 결과변수라고도 함

- 데이터프레임의 한 컬럼이 특성 또는 레이블이 됨

- 파라미터(Parameter)
    - 파라미터는 매개변수라고 하며, 모델이 학습하는 과정에서 자동으로 조정되는 값
    - 모델이 훈련 데이터로부터 학습하여 최적화하는 값으로, 학습 과정에서 갱신됨
    - 예 : 가중치(Weights), 편향(Bias)
- 하이퍼파라미터(Hyperparameter)
    - 모델을 훈련하기 전에 개발자가 설정해야 하는 값으로 외부 구성변수라고 함
    - 모델이 학습을 시작하기 전에 정의되고, 학습 중에 변경되지 않음
    - 예 : 분류의 개수(k), 학습률(Learning Rate), 배치 크기(Batch Size), 에포크 수(Epochs)

## 머신러닝 학습 방법
- 지도 학습(Supervised Learning)
    - 선생님이 학생을 가르치듯, 주어진 입력에 대해 **어떤 결과가 나오지 알고 있는 학습 데이터**를 이용해 주어진 데이터를 해석할 수 있는 모델을 만들고 이를 바탕으로 새로운 데이터를 추정하는 방법
    - 좋은 학습 결과를 얻기 위해서는 양질의 많은 데이터 확보 필요
    - 문제 유형 : 분류(Classification), 회귀(Regression)
    - 사용 사례: 이메일 스팸 필터링, 주식 가격 예측, 이미지 분류
- 비지도 학습(Unsupervised Learning)
    - 비지도 학습은 **주어진 데이터에 대한 결과가 없는 데이터**를 이용해 주어진 데이터에 내재된 패턴, 특성, 구조를 찾아서 학습하는 방법
    - 학습데이터만 넣어주면 스스로 학습하기 때문에 편리하지만 지도학습에 비해 결과가 항상 좋지 않음
    - 문제 유형 : 군집(Clustering), 차원축소(Dimension Reduction)
    - 사용 사례: 고객 세분화, 데이터 시각화, 이상 탐지
- 강화학습(Reinforcement Learning)
    - 에이전트의 동작이 적절한지에 대한 피드백을 반영하면서 학습
    - 에이전트가 특정환경에서 현재의 상태를 인식하여 보상이 최대화되는 행동을 수행하도록 하는 학습 방법
    - 문제 유형 : Q-러닝
    - 사용 사례: 자율주행, 게임 플레이, 로봇 제어

## 데이터 분할
- 훈련 오차(train error)가 낮다고 해서 항상 모델이 좋은 성능을 낸다고 판단할 수 없고, 새로운 데이터에서의 성능(test error/generalisation error)을 평가하는 것이 더 중요하므로 전체 데이터를 학습 데이터와 평가 데이터로 분할
    - 학습 데이터 : 모델을 학습시키기 위한 데이터셋
    - 평가 데이터 : 학습된 모델을 평가하기 위한 데이터셋
- 일반적으로 학습데이터 70%, 평가데이터 30%의 비율을 많이 이용
    - sklearn은 75%:25%가 기본값
- 데이터를 학습데이터, 검증데이터, 평가데이터로 3분할하여 사용하기도 함
    - 학습 50 : 검증 30 : 평가 20
- model_selection의 **train_test_split**을 이용하여 분할

```python
# 최적의 모델을 선택하는데 있어 어떻게 선택하냐? -> 학습 에러를 찾아서 나의 데이터를 잘 표현하는 모델찾고
#                                        일반화(expected error)찾아 일반화되는 모델을 찾는게 좋다.
# 그러므로 학습 에러가 낮다고 해서 좋은 성능 될 수 없다 / 평가했을 때 test error 도 보아야하기 때문에
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
```

```python
# 데이터분할
from sklearn.model_selection import train_test_split

x = np.arange(10) # test 데이터로 열개 가져옴

x_train, x_test = train_test_split(x, random_state =0)

# 모델 트레이닝 할 때 사용할 x_train
print(x_train) # train_test_split이 7개

# 모델을 테스트 할 때 사용할 x_test
print(x_test) # 3개로 분리 해줌
```

## 과소적합&과대적합
- 과소적합(Underfitting)
    - 정보를 충분히 학습하지 못해 학습데이터와 평가데이터에 대한 설명력 부족 (성능 저하)
- 과대적합(Overfitting)
    - 정보를 과도하게 학습해 일반화되어 학습 데이터에서는 높은 정확도를 보이지만, 평가 데이터나 학습하지 않은 데이터에 대해선 성능이 저하
- 과소적합과 과대적합은 모델의 복잡도에도 영향을 받지만 학습데이터의 품질에도 큰 영향을 받음
    - 보유한 데이터가 모델링을 위해 충분한 수준인지를 파악 후 머신러닝 기법 또는 하이퍼파라미터 튜닝과 같은 작업을 통해 문제 해결

```python
# 과소 적합 : train error / test error 모두 높음
# 과대 적합 : train error 낮음 / test error 만 높음

# 그래프에서 M 차원
# 세번 째 점까지는 under-fitting
# 마지막에서 두번 째 점부터 test error는 높아지고 train error 는 낮아짐 : over-fitting
# 그 사이는 적절한 fitting (적절한 k의 범주) 으로 볼 수 있겠다.

# 적절한 k를 구하는 것은 하이퍼파라미터튜닝이 중요하다!
```

## Bias(편향)&Variance(분산)
- Bias(편향)는 데이터 내에 있는 모든 정보를 고려하지 않음으로 인해, 지속적으로 잘못된 것을 학습하는 경향
    - Bias는 학습데이터를 바꿈에 따라서 알고리즘의 평균 정확도가 얼마나 많이 변하는지 보여줌
    - 예측값이 정답과 대체적으로 모두 멀리 떨어져있다면, "Bias가 높다"고 함
- Variance(분산)는 데이터 내에 있는 에러나 노이즈까지 잘 잡아내는 모델에 데이터를 학습시킴으로써 실제 정답과 연관없는 정보까지 학습하는 경향
    - Variance는 특정 입력 데이터에 대해 알고리즘이 얼마나 민감한지 나타냄
    - 예측값끼리의 값이 멀리 떨어져 있다면, "Variance가 높다"고 함

- 새로운 관측치와 각각의 학습데이터 사이의 거리를 전부 계산해야 하기 때문에 계산 시간이 오래 걸린다는 단점이 존재

## 거리 계산
- 최근접 이웃을 구하기 위해 거리 측정 방법 또는 유사도를 측정하는 방법을 활용
- 거리를 계산하는 다양한 방법이 존재하며, 일반적으로 유클리드 거리(Euclidean Distance)를 주로 이용
- 유클리드 거리
    - n개의 특성을 가진 두 데이터 p와 q의 거리 계산
    -
$dist(p,q) = \sqrt{(p_1 - q_1)^2 + \cdots + (p_n - q_n)^2}$
- 새로운데이터(평가데이터)를 분류하기 위해서는 기준 데이터(학습데이터)와의 가장 가까운 이웃 K에 대하여 거리 계산을 통해 평가데이터를 분류

#### [예제] 1NN
image.png

- 거리계산 결과
    -
$dist(A,q) = \sqrt{(5.1-4.9)^2 + (3.5-3.0)^2 + (1.4-1.4)^2 + (0.2-0.2)^2} = 0.538156$
    -
$dist(B,q) = \sqrt{(7.0-4.9)^2 + (3.2-3.0)^2 + (4.7-1.4)^2 + (1.4-0.2)^2} = 4.096339$
    -
$dist(C,q) = \sqrt{(6.3-4.9)^2 + (3.3-3.0)^2 + (6.0-1.4)^2 + (2.5-0.2)^2} = 5.338539$

- K=1이므로, 가장 가까운 이웃은 A이며, Blue범주를 가지고 있기 때문에 새로운 데이터로 Blue로 분류

### 3. 모델 학습
- KNN 모델을 neighbors에 구현되어 있으며, K는 n_neighbors로 지정
    - 분류 모델 : KNeighborsClassifier
    - 회귀 모델 : KNeighborsRegressor
- **fit 함수**를 이용하여 학습

```python
# 모델 학습
from sklearn.neighbors import KNeighborsClassifier # scikit-learn에서 KNN 분류 모델을 불러옵니다.
knn_model = KNeighborsClassifier(n_neighbors = 3)  # 몇개의 이웃을 가지고 할거냐 우리는 3개로 진행
knn_model.fit(X_train, y_train)                    # 학습 데이터(X_train)와 정답(y_train)을 이용해 모델을 학습
# 나온 이미지는 파이프라인
```

### 3. 모델 평가
- 모델은 **score 함수**를 통해 모델의 성능을 확인할 수 있음
    - 분류 모델인 경우 정확도(Accuracy), 회귀모델인 경우 R2를 이용
- 평가는 학습데이터와 평가데이터 모두 진행하며 학습데이터의 성능과 평가데이터의 성능 차이를 계산하여 모델의 성능을 최종 평가
    - 학습데이터의 예측 성능이 평가데이터의 예측 성능보다 훨씬 좋은 경우 과대적합이 발생한 것으로 추정

```python
# 모델 평가
print("학습 정확도/성능 : ", knn_model.score(X_train, y_train))
print("평가 정확도/성능 : ", knn_model.score(X_test, y_test))
```

### 4. 모델 예측
- **predict 함수**를 이용하여 예측 결과를 계산

    - 분류 모델인 경우 **predict_proba 함수를 이용하여 범주별 예측 확률을 계산**할 수 있음

```python
# 모델 예측
print("실제 : " , y_test)
print("예상 : ", knn_model.predict(X_test)) # X_test 값을 가지고 예상
```

### 5. 하이퍼파라미터 튜닝(K값의 선택)
- K는 KNN모델에서 가장 중요한 하이퍼파라미터
- 가장 좋은 K값을 찾기 위한 사전에 정의된 통계 또는 수학적 방법은 존재하지 않음
- 일반적으로 3~10 사이에서 주로 많이 결정하거나 학습데이터의 개수의 제곱근으로 설정
- 테스트를 통해 가장 좋은 성능을 보이는 K값을 찾는 것이 최선
    - 반복문을 이용하여 테스트
    - 하이퍼파라미터를 찾아주는 Tunner(GridSearch) 이용

```python
# 하이퍼파라미터 튜닝
## K를 1~10으로 설정하여 K값별 학습 및 평가 성능을 저장
# 그리드 서치는 다음시간에 하고 반복문먼저 해보겠다
k_range = range(1,11)
train_scores = []
test_scores = []

for k in k_range :
    knn = KNeighborsClassifier(n_neighbors = k)
    knn.fit(X_train, y_train) # 모델 학습
    train_scores.append(knn.score(X_train, y_train)) # 학습 성능과
    test_scores.append(knn.score(X_test, y_test))    # 평가 성능 계산해 배열에 저장 어떤게 제일 좋은지 찾아볼 수 있음

print(train_scores)
print(test_scores)
```

```python
# 결과 시각화
plt.plot(k_range, train_scores, label = 'train accuracy')
plt.plot(k_range, test_scores, label = 'test accuracy')
plt.xticks(k_range)
plt.legend()
plt.show()

# 최적은 4, 5, 7, 8, 10
# 1은 둘다 높은 성능을 가지고 있지만, 우선 1은 train 정확도가 test보다 높다.
# 이런 경우에는 과대적합의 가능성을 가지고 있다.
```

```python
# 최적의 K값 찾기
# 하이퍼파라미터 튜닝
## K를 1~10으로 설정하여 K값별 학습 및 평가 성능을 저장
k_range = range(1,11)
train_scores = []
test_scores = []

for k in k_range :
    knn = KNeighborsClassifier(n_neighbors = k)
    knn.fit(X_train, y_train) # 모델 학습
    train_scores.append(knn.score(X_train, y_train)) # 학습 성능과
    test_scores.append(knn.score(X_test, y_test))    # 평가 성능 계산해 배열에 저장 어떤게 제일 좋은지 찾아볼 수 있음

print(train_scores)
print(test_scores)
```

```python
# weights="uniform" , weights="distance" 로 무게를 줄 수 있음
# uniform : 거리에대한 가중치를 두지 않음
# distance : 가중치를 거리로 준다. (가까운 값일 수록 가중치를 더 줌)

# 가중치를 안 주는 거면 유클리드 거리 그대로 계산하는 거고
# 주는 거면 가중치에 해당되는 거리 계수가 추가되는 거 같네요

# 그렇다고 knn이 불균형을 해소해주지는 않기 때문에 오버/언더샘플링으로 불균형을 먼저 해결하는 방식이 더 많이 사용된다.
```

```python
# weights = 'distance'

k_range = range(1,11)
train_scores = []
test_scores = []

for k in k_range :
    knn = KNeighborsClassifier(n_neighbors = k, weights = 'distance')
    knn.fit(X_train, y_train) # 모델 학습
    train_scores.append(knn.score(X_train, y_train)) # 학습 성능과
    test_scores.append(knn.score(X_test, y_test))    # 평가 성능 계산해 배열에 저장 어떤게 제일 좋은지 찾아볼 수 있음

print(train_scores)

# 최적의 K값 찾기 시각화
plt.plot(k_range, train_scores, label = 'train accuracy')
plt.plot(k_range, test_scores, label = 'test accuracy')
plt.xticks(k_range)
plt.legend()
plt.show()
```
