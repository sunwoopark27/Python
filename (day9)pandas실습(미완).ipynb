```python
# csv 파일읽기
# 주어진 파일의 encoding을 자동으로 검사한 후 pandas를 통해 파일을 읽음.
import chardet

def read_csv_file(file_path, index_col=0):
    try:
        # 1. 파일을 바이너리 모드('rb')로 읽고, 인코딩 감지
        with open(file_path, 'rb') as f :
            result = chardet.detect(f.read())

        # 2. 감지된 인코딩을 사용하여 CSV 파일 읽기
        df = pd.read_csv(file_path,
                         encoding=result['encoding'],
                         index_col=index_col)

    except Exception as e:
        print(f"Error:{e}")

    return df # 성공적으로 읽은 DataFrame 반환
```

```python
# 인당 GRDP 파일읽기
grdp = pd.read_csv('/content/drive/MyDrive/1인당_GRDP_시도_15_23.csv',encoding='EUC-KR'
                    ,index_col=0)
grdp
```

```python
# 2023 p) 컬럼 이름 변경
grdp.rename(columns={'2023 p)':'2023'}, inplace=True)
grdp.head()
```

```python
# 연도별 GRDP가 높은 지역
grdp.sum(axis=1).sort_values(ascending=False).head(1)
```

```python
### 풀이
# 연도별 GRDP가 높은 지역
                    # 1) 열방향으로 최대값의 인덱스 찾기
highest = grdp.index[np.argmax(grdp, axis=0)]
      # 2) 반환된 인덱스 번호를 이용해 원래 DataFrame의 인덱스 레이블 찾기

lowest = grdp.index[np.argmin(grdp, axis=0)]
highest, lowest
#grdp.idxmax(axis=0), grdp['2023'].idxmin(axis=0)
```

```python
# 2020년 대비 2023년 증감율
grdp['증감율'] = (grdp['2023'] - grdp['2020'])/grdp['2020']
grdp['증감율'].sort_values(ascending=False)
```

```python
# 사회조사 지표 파일 읽기
social = pd.read_csv('/content/drive/MyDrive/사회조사지표_시도_2023.csv', index_col=0)
social.head()

# 풀이
# social_path = '/content/drive/MyDrive/LikeLion/!수업자료/사회조사지표_시도_2023.csv'
# social = read_csv_file(social_path)
# social.head(2)
```

```python
# 인구통계 파일읽기
population_path = '/content/drive/MyDrive/인구통계_시도_2023.csv'
population = read_csv_file(population_path)
population.head(2)
```

```python
# 인구통계의 컬럼명 변경
population = population.rename_axis('행정구역별')
# 인구통계의 인덱스명 변경
population = population.rename(index={'전라북도':'전북특별자치도'})
population
```

```python
# 데이터 병합 (좌우)
df = pd.concat([grdp[['2023','증감율']], social, population], axis=1)
df.head(2)
```

```python
# 2023년 GRDP의 컬럼명 변경
df.rename(columns={'2023':'인당GRDP'}, inplace = True)
# df의 결측치 및 데이터 타입 확인
df.info()
```

```python
# 범죄발생건수 숫자형으로 변환
df.loc['서울특별시','범죄발생건수'] = 277213
```

```python
# 범죄발생건수 숫자형 변환
df.loc['범죄발생건수'] = pd.to_numeric(df['범죄발생건수'])
df['범죄발생건수'].head()
```

```python
# 변수간 상관분석
df_corr_matrix = df.corr()
df_corr_matrix
```

```python
# 삶의 만족도와 강한 상관관계 확인
df.corr()['삶의만족도'].sort_values(ascending=False).head()

## 풀이
high_relevance = abs(df_corr_matrix['삶의만족도'])>=0.7
df_corr_matrix[high_relevance].index
```

```python
# 인구밀도와 강한 상관관계 확인
high_relevance = abs(df_corr_matrix['총인구 (명)'])>=0.7
df_corr_matrix[high_relevance].index
```

```python
# 변수에 대한 상관관계가 높은 변수명 출력 함수
def high_relevance(df, label, threshold=0.7):
    df_corr_matrix = df.corr()
    high_relevance = abs(df_corr_matrix[label])>=threshold
    relevance_list = list(df_corr_matrix[high_relevance].index)
    relevance_list.remove(label) #변수(자기자신) 삭제
    return relevance_list
```

```python
high_relevance(df,'삶의만족도')
```
