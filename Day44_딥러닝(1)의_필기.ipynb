# 머신러닝과 딥러닝
- 머신러닝(Machine Learning)은 명시적인 프로그래밍 없이 컴퓨터가 데이터를 통해 학습하고, 스스로 규칙을 도출하여 문제를 해결하는 인공지능(AI)의 하위 분야
- 딥러닝(Deep Learning)은 인공지능과 머신러닝의 하위 분야로 인공신경망(Artificial Neural Networks)을 기반으로 한 모델을 사용하여 데이터를 학습하고 복잡한 패턴을 인식하는 기술
    - 딥러닝이라는 이름은 인공신경망이 여러 층을 가짐으로써 복잡한 데이터의 구조를 잘 이해하고 처리할 수 있다는 점에서 유래
- 딥러닝은 특히 이미지 인식, 음성 인식, 자연어 처리와 같은 문제를 해결하는 데 뛰어난 성능을 보임
    - 딥러닝은 계층적 표현 학습과 자동화된 특징 추출을 통해 복잡한 데이터를 효과적으로 분석하며 대규모 데이터 처리와 전이 학습 능력 덕분에 높은 성능을 발휘
- 딥러닝은 사람의 뇌 신경망을 모방한 인공신경망을 사용하며 인공신경망은 여러 개의 노드(Node) 또는 뉴런(Neuron)으로 이루어져 있으며, 이들 뉴런은 서로 연결되어 계층(Layer)을 형성함
    - 인공신경망은 입력층(Input Layer), 은닉층(Hidden Layer), 출력층(Output Layer)으로 구성됨

# Tensor

## Tensor 생성
- `tensorflow.constant()`와 `tensorflow.Variable()`를 사용하면, 기존의 스칼라, 리스트, NumPy 배열 등으로부터 Tensor를 생성할 수 있음
    - 입력 데이터를 복사하여 Tensor를만들기 때문에 입력 데이터가 변경되어도 Tensor는 영향을 받지 않음
    - constant()는 불변 텐서, Variable()는 수정 가능한 텐서
    - 입력 데이터의 타입에 맞추어 Tensor의 데이터 타입을 설정함
        - dtype 인수를 통해 명시적으로 설정할 수도 있음
- 특정 값으로 초기화된 Tensor를 생성하는 다양한 함수를 제공
    - `zeros(shape)` : 모든 요소가 0인 Tensor를 생성
    - `zeros_like(input)` : input과 동일한 크기와 데이터 타입을 가지면서 모든 요소가 0인 Tensor 생성
    - `ones(shape)` : 모든 요소가 1인 Tensor를 생성
    - `ones_like(input)` : input과 동일한 크기와 데이터 타입을 가지면서 모든 요소가 1인 Tensor 생성
    - `fill(dims, value)` : 주어진 값(value)으로 채워진 Tensor를 생성
    - `eye(num_rows, num_columns)`: 단위 행렬(identity matrix) 또는 대각선이 1로 채워진 행렬을 생성
    - `range(start, limit, delta)`: 주어진 범위 내에서 일정 간격으로 값이 증가하는 1D Tensor를 생성
    - `linspace(start, end, num)`: 시작점과 끝점을 포함하여 일정한 간격으로 나뉜 값을 갖는 Tensor를 생성
    - `random.uniform(shape, minval=0, maxval=1)` : [minval, maxval)범위에서 균일하게 분포된(random uniform distribution) 난수로 채워진 Tensor를 생성
    - `random.normal(shape, mean=0, stddev=1)` : 평균이 mean, 표준편차가 stddev인 정규 분포(normal distribution)를 따르는 난수로 채워진 Tensor를 생성
    - `random.shuffle(value)` : value를 랜덤하게 섞은 Tensor를 생성


````python
# 리스트를 tensor로 변환
a = [1, 2, 3]
tf.constant(a)

# 실수 데이터 타입으로 tensor 변환
tf.constant(a, dtype='float64')
````

## 데이터 선택
- 인덱싱(indexing)과 슬라이싱(slicing)을 지원
    - 특정 요소에 접근하거나 일부를 선택할 수 있으며, 복잡한 데이터 조작을 수행할 수 있음
    - 변수명[인덱스] 형태로 특정 요소에 접근 할 수 있음
- 기본 인덱싱은 특정 위치의 요소를 선택하는 가장 단순한 방법으로 Python 리스트와 매우 유사하게 사용
    - 위치 인덱스는 0부터 시작하며 마지막의 데이터는 -1로 접근 가능
- 다차원 배열의 경우 쉼표(,)를 기준으로 각 차원별 인덱스를 지정할 수 있음
- 슬라이싱은 Tensor의 특정 부분을 선택하여 부분 Tensor를 반환하며 콜론(:)을 사용하여 범위를 지정
    - 변수[시작인덱스:끝인덱스] 형태로 사용하며 끝 인덱스는 포함되지 않음
    - 시작 인덱스를 생략 하는 경우 처음부터 데이터를 가지고 옴
    - 끝 인덱스를 생략 하는 경우 끝까지 데이터를 가지고 옴
- 조건 인덱싱(Boolean Indexing)은 Tensor의 요소를 조건에 따라 선택할 때 사용
    - 조건을 만족하는 요소만 선택하여 새로운 Tensor를 생성
    - tensor의 비교 연산은 각 요소별 비교를 수행하며 True, False 값을 가지는 Tensor를 반환함
- 팬시 인덱싱(Fancy Indexing)은 `tensorflow.gather(params, indices, axis)`함수를 사용하여 특정 인덱스를 선택하는 방법
    - 비연속적인 요소들을 선택할 수 있음

````python
# 1D 텐서 생성
tensor_1d = tf.Variable([10, 20, 30, 40, 50])

# 첫번째 요소 선택
tensor_1d[0].numpy() # .numpy()를 통해 numpy값만 가져올 수 있다.
````

````python
# 마지막 요소 선택
tensor_1d[-1].numpy()
````

````python
# 마지막행, 마지막열 요소 선택
tensor_2d[-1,-1].numpy()
````

````python
# 1D 텐서의 슬라이싱
# 인덱스 1부터 3까지의 요소 선택
tensor_1d[1:4]

# 처음부터 인텍스 3까지의 요소 선택
tensor_1d[:3]

# 인덱스 2부터 끝까지 요소 선택
tensor_1d[2:]
````

````python
# 2D 텐서의 슬라이싱
# 첫번째 행 선택
tensor_2d[0, :]

# 첫번째 열 선택
tensor_2d[:,0]

# 첫번째와 두번째 행 선택
tensor_2d[:2,:]
````

````python
# 1D 텐서에서 30보다 큰 요소만 선택
print(tensor_1d[tensor_1d > 30].numpy())

# 1D 텐서에서 짝수인 요소만 선택
print(tensor_1d[tensor_1d % 2 ==0].numpy())

# 2D 텐서에서 5보다 큰 요소만 선택
print(tensor_2d[tensor_2d > 5].numpy())
````

````python
# 1D 텐서에서 인덱스 0, 2, 4 위치의 요소 선택
tf.gather(tensor_1d, [0,2,4]).numpy()
````

````python
# 2D 텐서에서 첫 번째와 세 번째 행 선택
tf.gather(tensor_2d,[0,2], axis=0).numpy()
````

````python
# 2D 텐서에서 (0, 1)과 (2, 2) 위치의 요소 선택
tf.gather_nd(tensor_2d, [[0,1] , [2,2]]).numpy()
````

## Tensor 연산
- Tensor 연산은 수학적, 논리적, 변환 연산을 포함하여 다양한 종류의 연산을 지원
    - 기본 산술 연산 : 덧셈, 뺄셈, 곱셈, 나눗셈 등이 포함되며 이러한 연산은 요소별(element-wise)로 수행
        - 브로드캐스팅 : 크기가 다른 Tensor 간 연산을 지원하는 기능으로 작은 Tensor가 큰 Tensor와의 연산에서 자동으로 확장
    - 행렬 연산 : 행렬 곱셈, 전치(transpose), 역행렬(inverse)
    - 집계 연산 : Tensor의 요소들에 대해 합계(sum), 평균(mean), 최대값(max), 최소값(min) 등

````python
# Tensor 생성
x = tf.constant([1, 2, 3])
y = tf.constant([4, 5, 6])

# 요소별 덧셈
print(tf.add(x, y))
print(x+y)

# 요소별 뺄셈
print(tf.subtract(x, y))
print(x-y)

# 요소별 곱셈
print(tf.multiply(x, y))
print(x*y)

# 요소별 나눗셈
print(tf.divide(x, y))
print(x/y)
````

````python
# 브로드캐스팅
# Tensor 생성
x = tf.constant([[1, 2, 3], [4, 5, 6]])
y = tf.constant([1, 2, 3])

# 브로드캐스팅을 통한 요소별 덧셈
# 두 행에 모두 더해진 것을 볼 수 있다.
print(tf.add(x, y))
print(x+y)
````

````python
# 행렬 연산
# Tensor 생성
x = tf.constant([[1, 2, 3], [4, 5, 6]])      #  (2x3 행렬)
y = tf.constant([[7, 8], [9, 10], [11, 12]]) # (3x2 행렬)

# 행렬 곱셈
print(tf.matmul(x, y))
print(x@y)

# x의 전치 행렬 # 여기에는 .T 적용 안됨
print(tf.transpose(x))

# 역행렬(단위행렬이 나오게 하는 행렬) # 정사각행렬에만 가능
t = tf.constant([[1, 2], [3, 4]], dtype='float32')
print(tf.linalg.inv(t))
````

````python
# 집계 연산
# Tensor 생성
x = tf.constant([[7, 1, 3],[2, 8, 6],[4, 5, 9]])

# 전체 합계
print(tf.reduce_sum(x))

# 열별 합계
print(tf.reduce_sum(x, axis=0))

# 평균
print(tf.reduce_mean(x))

# 행별 평균
print(tf.reduce_mean(x, axis=1))

# 최대값
print(tf.reduce_max(x))

# 최소값
print(tf.reduce_min(x))

# # 최대값의 위치
# print(tf.argmax(x)) # 전체의 최대값 위치를 가져와야하는데 안되고 있음

# 열별 최대값의 위치
print(tf.argmax(x, axis=0))

# 행별 최소값의 위치
print(tf.argmin(x, axis=1))
````

## Tensor 재구조화(reshaping)
- Tensor의 재구조화는 데이터의 실제 값은 변경하지 않으면서 텐서의 형태(shape)를 변경하는 작업
    - `reshape()` 함수는 텐서의 모양을 변경
    - 텐서가 연속적이지 않아도 동작하며 필요에 따라 데이터 복사본을 생성
- tf.transpose()의 `perm`인자를 활용하여 텐서의 모든 차원의 순서를 지정한 대로 재배열



````python
# 4차원 텐서 생성 : (10, 28, 28, 3)
x_4d = tf.random.normal([10, 28, 28, 3])  # (10, 28, 28, 3)

# 텐서를 (10, 28 * 3 * 3)으로 변환, 즉 2차원 텐서로 변경
x_2d = tf.reshape(x_4d, (10, -1))
# -1은 나머지 차원을 자동으로 계산하게 하여 (10, 2352)으로 변경
print(x_2d.shape)

# 차원의 순서를 (10, 3, 28, 28)로 재배열 (4차원을 2차원으로 이동)
x_p = tf.transpose(x_4d, perm=[0,3,1,2])
print(x_p.shape)
````

## 단층 퍼셉트론의 한계
- 기본적으로 퍼셉트론은 단층 퍼셉트론 (Single Layer Perceptron) 구조로 입력층과 출력층으로 구성됨
    - 단층 퍼셉트론은 선형 분리만 가능하다는 문제점이 존재
image.png

- 비선형 분리를 위해 은닉층이 추가된 다층 퍼셉트론 (Multi Layer Perceptron) 구조를 사용

    image.png

````python
# 단층 퍼셉트론 구성
def perceptron(X=X, y=y, learning_rate=0.1, epochs=10000):
    input_dim = X.shape[1] # 크기를 가져옴 # x,shape -> (샘플 수,feature수(속성변수의 개수))
    output_dim = y.shape[1] # 크기를 가져오기 위해 [1]

    # 가중치와 편향 초기화
    weights = tf.Variable(tf.random.normal([input_dim, output_dim]))
    bias = tf.Variable(tf.random.normal([output_dim]))
    # 초기화 시에 어떤 값을 초기화하냐에 따라 학습의 방향이나 속도가 달라짐

    # 활성화 함수 (Sigmoid)
    def sigmoid(x):
        return 1 / (1 + tf.exp(-x))

    # 훈련 과정
    for epoch in range(epochs):
        # 기존값을 예측하고 그것에 대한 손실 계산을 해줌
        # 경사하강법을 사용해 가중치와 편향 업데이트

        # 순전파
        # 계산한 값
        z = tf.matmul(X, weights) + bias  # X와 가중치를 곱하고 편향 추가
        predictions = sigmoid(z)  # 활성화 함수 적용

        # 손실 계산 (평균 제곱 오차)
        loss = tf.reduce_mean(tf.square(predictions - y))

        # 경사 하강법을 사용한 가중치와 편향 업데이트
        with tf.GradientTape() as tape:
            tape.watch(weights)
            tape.watch(bias)
            # 순전파
            z = tf.matmul(X, weights) + bias
            predictions = sigmoid(z)
            loss = tf.reduce_mean(tf.square(predictions - y))

        # 역전파 (기울기 계산)
        gradients = tape.gradient(loss, [weights, bias])

        # 경사 하강법 업데이트
        weights.assign_sub(learning_rate * gradients[0])
        # weight = weight(기존) - learning_rate * 미분값
        bias.assign_sub(learning_rate * gradients[1])

        # 2000번 반복시마다 Loss 출력
        if (epoch + 1) % 2000 == 0:
            print(f'Epoch [{epoch+1}/10000], Loss: {loss.numpy():.4f}')

    # 결과 확인
    z = tf.matmul(X, weights) + bias # (X * weight) + bias
    predictions = sigmoid(z) # 활성함수 적용
    prediction_label = tf.cast(predictions >=0.5, float)
    print(f"Predicted Prob: {predictions.numpy()}")
    print(f"Predicted Labels: {prediction_label}")
````

````python
# 순전파를 이용해 예측값을 계산하고 이것으로 인해 Loss(1/n(예측-실제)^2 : meansquarederror라고 가정)가 나옴
# Loss를 최소화 시키기 위한 w, b 파라미터 업데이트
# -> 이렇게 업데이트할 때 사용하는 값을 경사하강법으로 구함 : 로스를 이용하는 방법이므로 역전파
# 업데이트 후에 새로운 w,b가 생기면 처음부터 반복(순전파-> 예측계산 -> Loss -> b,w 업데이트 -> 역전파..)
# 반복 -> epoch
# 반복해 최종 w, b를 기반으로 예측값 계산
````

## 심층신경망
- 심층 신경망(Deep Neural Network, DNN)은 딥러닝(Deep Learning)에서 사용하는 주요 모델 중 하나
    - 심층 신경망은 입력 데이터로부터 복잡한 패턴을 학습하여, 다양한 예측 및 분류 작업을 수행할 수 있음
- 심층 신경망은 인간의 뇌 구조를 모방한 인공 신경망(Artificial Neural Network, ANN)의 일종으로입력층(Input Layer), 여러 개의 은닉층(Hidden Layer) 출력층(Output Layer)을 가진 신경망을 의미함
    - 각각의 층(Layer)은 뉴런(Neuron)이라고 불리는 계산 단위로 구성되며 뉴런은 가중치(Weight)와 편향(Bias)이라는 매개변수를 사용하여 데이터를 처리함
    - 입력층: 데이터를 받아들이는 층으로 예를 들어, 이미지 인식에서는 이미지의 픽셀 값을 입력층에 전달
    - 은닉층: 입력 데이터에서 특징을 추출하고, 학습을 통해 패턴을 파악하는 역할을 하는 층으로 심층 신경망에서는 이러한 은닉층이 여러 개 존재하며, 각 층은 이전 층의 출력을 받아 처리
    - 출력층 : 최종 예측 결과를 출력하는 층으로 예를 들어, 이미지 분류 문제에서는 출력층에서 각 클래스에 대한 확률을 반환

### keras.layer.Dense
- keras.layer.Dense는 Dense Layer 또는 Fully Connected Layer라고도 불리며, 가장 기본적인 신경망 레이어로 입력 데이터에 대해 선형 변환을 적용하고, 그 결과를 출력으로 내보냄
    - 신경망의 각 뉴런이 이전 레이어의 모든 뉴런과 연결되어 있기 때문에 완전 연결(fully connected)이라고 불림

```
tf.keras.layers.Dense(units, activation, use_bias, kernel_initializer, bias_initializer)
```
- `units`: 출력 차원의 크기 (즉, 뉴런의 수)
- `activation` : 활성화함수 지정 (기본값 None)
- `use_bias`: 편향을 사용할지 여부 (기본값 True)
- `kernel_initializer`: 가중치 행렬의 초기화 방법 (기본값 glorot_uniform)
    - glorot_uniform : 입력과 출력의 수에 따라 조정, ±limit의 범위의 균등분포에서 샘플링 <br>
$limit = \sqrt{\frac{6}{\text{input_dim}+\text{output_dim}}}$
    - glorot_normal : 평균 0, 표준편차가
$\sqrt{\frac{2}{\text{input_dim}+\text{output_dim}}}$ 인 정규분포에서 샘플링
    - he_uniform : 입력의 수에 따라 조정, ±limit의 범위의 균등분포에서 샘플링 <br>
$limit = \sqrt{\frac{6}{\text{input_dim}}}$
    - he_normal : 평균이 0이고, 표준편차가
$\sqrt{\frac{2}{\text{input_dim}}}$ 인 정규분포에서 랜덤 초기화
- `bias_initializer` : 편향의 초기화 방법 (기본값 zeros)
    - zeros : 0으로 초기화


#### keras를 사용한 다층 퍼셉트론 구현

````python
# keras 라이브러리 임포트
from tensorflow.keras import layers, models
````

````python
# Sequential 모델을 사용하여 간단한 다층 퍼셉트론 모델 정의
model = models.Sequential([
    layers.InputLayer(shape=(X.shape[1],)),  # 입력층
    layers.Dense(64, activation='relu', kernel_initializer='he_normal'), # 첫번째 은닉층, 64개의 뉴런
    layers.Dense(32, activation='relu'),  # 두 번째 은닉층, 32개의 뉴런
    layers.Dense(1, activation='sigmoid')  # 출력층, 이진 분류의 경우 'sigmoid' 활성화 함수
])
````

````python
# 모델 컴파일
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
````

````python
# 모델 학습
model.fit(X, y, epochs=100)
````

````python
# 예측
predictions = model.predict(X)
predicted_labels = (predictions >= 0.5).astype(float)  # 0.5 이상은 1, 미만은 0으로 변환

# 결과 출력
print("Predicted Labels: ", predicted_labels)
````

### 순전파
- 순전파(Forward Propagation)는 인공 신경망에서 입력 데이터를 받아 출력 결과를 계산하는 과정
- 신경망이 학습 중에 주어진 입력 데이터에 대해 어떤 예측을 수행할지를 결정하는 중요한 역할
- 입력층에서 원시 데이터가 신경망으로 들어오며 이 데이터는 벡터 형태로 표현되며, 각 원소는 입력층의 노드에 해당됨
- 은닉층에서 각 입력값은 가중치(weight)와 곱해지며, 모든 가중된 입력 값들의 합을 계산하여 활성화 함수에 전달
    - 가중치(Weight): 각 입력에 할당된 가중치는 입력의 중요도를 의미함. 학습 과정에서 가중치는 조정됨
    - 편향(Bias): 편향은 뉴런의 출력에 추가되는 값으로, 모델이 더 유연하게 학습할 수 있게 함
- 활성화 함수는 가중합 결과를 비선형적으로 변환하여, 신경망이 복잡한 데이터 패턴을 학습할 수 있도록 함

    - 예측 값은 분류 문제의 경우 확률로 표현될 수 있으며, 회귀 문제에서는 연속적인 값으로 나타날 수 있음

### 활성화 함수
- 활성화(Activation) 함수는 인공신경망에 비선형성을 부여하기 위한 도구로 선형 결합 결과(가중치 곱의 합)가 비선형 활성화 함수를 통과하면 비선형적인 결과로 변환
    - 각 층이 거듭되면서 누적되는 비선형적인 결과들이 모델에 비선형적인 특징을 구분할 수 있는 능력을 부여함
- 은닉층에서는 ReLU함수를 가장 많이 사용하며 모델 구성에 따라 Leaky ReLU, ELU 등을 사용하기도 함
    - Sigmoid 함수는 현재는 주로 출력층에서 사용됨

image.png

````python
import matplotlib.pyplot as plt
# 활성화 함수 그래프 그리는 함수 정의
def plot_activation_function(activation_function):
    # x 값 생성 (-10에서 10까지 1000개의 점 생성)
    x = tf.linspace(-10, 10, 1000)
    y = activation_function(x)  # 활성화 함수 적용

    # 그래프 그리기
    plt.figure(figsize=(4, 3))
    plt.plot(x.numpy(), y.numpy(), label=str(activation_function.__name__))
    plt.title(f'Activation Function: {activation_function.__name__}')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.grid(True)
    plt.legend()
    plt.show()
````

#### Sigmoid
- Sigmoid 함수는 이진 분류 문제에서 주로 사용되는 활성화 함수로 입력 값을 0과 1 사이로 압축 <br>
$sigmoid(x) = \frac{1}{1 + e^{-x}}$
- 출력이 0과 1 사이에 존재하므로 확률적인 해석이 가능하며 순전파와 역전파에서 수학적으로 간단함
    - 이진 분류 문제에서 출력 값을 0 또는 1로 정규화하는 데 자주 사용
- 출력이 극단적으로 작거나 크면, 미분 값이 매우 작아지고 이로 인해 층이 깊어질수록 기울기가 급격히 줄어드는 기울기 소실
(Gradient Vanishing) 문제가 발생
    - Sigmoid 함수의 미분값이 최대 0.25로 n개의 Sigmoid 층을 거치면 최종 기울기는
$0.25^n$로 기울기가 소실됨

````python
# 시그모이드 함수 시각화
def sigmoid(x):
    return 1 / (1 + tf.exp(-x))

plot_activation_function(sigmoid)
plot_activation_function(tf.keras.activations.sigmoid)
````

#### Softmax
- 다중 클래스 분류 문제에서 주로 사용되는 활성화 함수로 입력값을 0과 1 사이의 값으로 압축<br>
$softmax(x_i) = \frac{e^{x_i}}{\sum_{i=1}^{C}{e^{x_i}}}$
- 다중 클래스 분류 문제에서 출력 값을 0 또는 1로 정규화하는 데 자주 사용
- Sigmoid와 동일하게 출력 값이 매우 큰 값이나 작은 값일 경우, 기울기 소실(Gradient Vanishing) 문제 발생


````python
# 소프트맥스 함수 시각화
def softmax(x):
    e_x = tf.exp(x - tf.reduce_max(x, axis=0))
    return e_x / tf.reduce_sum(e_x, axis=0)

plot_activation_function(softmax)
plot_activation_function(tf.keras.activations.softmax)
````

#### Tanh (Hyperbolic Tangent)
- Tanh 함수는 순환신경망의 은닉층에서 주로 사용되는 활성화 함수로 입력 값을 -1과 1 사이로 압축 <br>
$tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$
- Sigmoid보다 출력 범위가 넓고, 중심이 0이기 때문에 RNN, LSTM 등에서 활성화 함수로 자주 사용
- 중심이 0에 있어 Sigmoid 보다 기울기 소실이 덜하지만, 극단적인 출력값에서는 여전히 문제가 발생
- tanh의 미분값은 출력값이 0에 가까울수록 최대 1, ±1에 가까울수록 0에 가까워짐

````python
# Tanh 활성화 함수 구현 및 시각화
def tanh(x):
    return (tf.exp(x) - tf.exp(-x)) / (tf.exp(x) + tf.exp(-x))

plot_activation_function(tanh)
plot_activation_function(tf.keras.activations.tanh)
````

#### ReLU(Rectified Linear Unit)
- ReLU는 양의 입력에 대해 그대로 출력하고, 음의 입력에 대해 0을 출력하는 비선형 활성화 함수 <br>
$ReLU(x) = max(0,x)$
- 딥러닝에서 가장 널리 사용되는 활성화 함수로, 대부분의 신경망에서 기본 활성화 함수로 사용
- 기울기 소실 문제가 적고, 양수 구간에서 기울기가 1로 일정하여 학습이 빠르고 계산이 단순하여 효율적
- 입력 값이 0보다 작을 경우 기울기가 0이 되어 뉴런이 학습에 참여하지 못하는 문제(dying ReLU)가 발생할 수 있음

````python
# ReLU 활성화 함수 구현 및 시각화
# ReLU 활성화 함수 구현 및 시각화
def relu(x):
    return tf.maximum(x, 0)

plot_activation_function(relu)
plot_activation_function(tf.keras.activations.relu)
````

#### Leaky ReLU
- Leaky ReLU는 ReLU의 변형으로, 음의 입력에 대해서도 작은 기울기를 가지도록 개선된 활성화 함수 <br>
$Leacky ReLU(x) = max(\alpha x,x)$
- 음의 입력에 대해서도 작은 기울기를 가지므로 뉴런이 죽는 문제를 완화하며 ReLU보다 안정적인 학습을 제공
- 기본적으로 α=0.01이 자주 사용되지만, 이 값은 데이터에 따라 조정이 필요하며 적절하지 않은 경우 성능 저하의 가능성이 있음

````python
# Leaky ReLU 활성화 함수 구현 및 시각화
def leaky_relu(x, alpha=0.01):
    return tf.maximum(x, alpha * x)

plot_activation_function(leaky_relu)
plot_activation_function(tf.keras.activations.leaky_relu)
````

#### ELU (Exponential Linear Unit)
- ELU는 ReLU와 유사하지만, 음의 입력에 대해 지수 함수를 사용하여 출력이 0에 수렴하도록 설계된 활성화 함수 <br>
$ELU(x) = f(x) =
\begin{cases}
x & \text{if } x > 0 \\
α(e^x -1) & \text{if } x \le 0
\end{cases}$
- 음의 영역에서도 기울기가 존재하므로 뉴런이 죽는 문제를 방지하며 출력이 0에 가까워져 학습 안정성이 높아지고, 경사 하강법의
수렴 속도가 개선
- ReLU에 비해 더 정교하게 음수 영역을 처리하므로, 비선형성을 더 효과적으로 학습할 수 있지만 ReLU에 비해 계산이 복잡하며,
특히 음수 구간에서 지수 함수를 계산하므로 계산 비용이 증가

````python
# ELU 활성화 함수 구현 및 시각화
def elu(x, alpha=1.0):
    return tf.where(x > 0, x, alpha * (tf.exp(x) - 1))

plot_activation_function(elu)
plot_activation_function(tf.keras.activations.elu)
````

### 기울기 소실(Vanishing Gradient)
- 신경망이 매우 깊을 경우, 역전파 과정에서 기울기가 점점 작아져서 최종적으로 가중치 업데이트에 거의 영향을 주지 않게 되는 문제로 학습이 매우 느려지거나 아예 멈추게 할 수 있음
- Sigmoid나 Tanh 같은 활성화 함수는 입력 값이 너무 크거나 작아질 때 출력 값이 극단적으로 0에 가까워지는 특징이 존재
    - Sigmoid 함수의 출력이 0.999나 0.001처럼 극단적일 때, 다음 층으로 전달되는 정보가 약화
- 반대로 기울기가 매우 커져서 가중치가 불안정하게 되는 기울기 폭발(Exploding Gradient) 문제도 발생할 수 있음
- 학습률이 너무 크면 학습이 불안정해지고, 너무 작으면 학습이 너무 느리게 진행되며 적절한 학습률을 선택하는 것이 중요함

### 손실 계산
- 출력층에서 얻어진 예측 값과 실제 값(레이블) 사이의 차이를 손실 함수(Loss Function)를 사용해 계산
    - 모델이 얼마나 정확하게 예측하고 있는지를 수치적으로 표현
    - 손실 값이 클수록 모델의 예측이 부정확하다는 것을 의미함
- 모델은 학습 과정에서 손실 값을 최소화하는 방향으로 가중치와 편향을 업데이트 함
    - 모델의 학습 목적(Objective)은 손실/비용을 최소화 하는 것
- 손실 함수는 문제의 유형(예: 회귀, 분류)에 따라 다를 수 있음
    - 회귀에서는 평균 제곱 오차(MSE, Mean Squared Error), 평균 절대 오차(MAE, Mean Absolute Error)등 사용
    - 분류 문제에서는 주로 교차 엔트로피(Cross-Entropy)를 사용 <br>
$Cross Entropy = - \frac{1}{N}\sum_{n=i}^N\sum_{i=1}^C y_{ni}\log(\hat{y}_{ni})$
        - 이진 교차 엔트로피 (Binary Cross-Entropy) : 교차 엔트로피를 이진 분류에 적합한 형태로 변형한 수식
$Binary Cross Entropy  = - \frac{1}{N}\sum_{n=i}^N[y_n \log(\hat{y}_{n}) + (1 - y_n) \log(1 - \hat{y}_{n})]$


모델을 컴파일을 할 때 들어가는 3가지의 값
- optimizer : 어떤 방식으로 업데이트 할건지
- Loss
- metrics : 모델의 성능을 평가하기 위하

### 경사 하강법 (Gradient Descent)
- 경사 하강법은 일반적으로 최적의 파라미터를 찾기(최적화 문제) 위해 사용하는 방법
    - 기본적인 접근 방법은 임의의 파라미터를 대입하여 예측 값을 생성하고, 차이(손실)를 확인한 후 파라미터를 조정하는 것
    - 이 과정에서 임의의 파라미터를 입력하는 것은 매우 비효율적이고, 최적 값을 찾을 가능성이 매우 낮음
    - 대입할 파라미터를 계산을 통해 찾는 방법 중 대표적인 방법이 경사 하강법
- 경사 하강법은 미분 값을 이용하여 파라미터를 업데이트 하는 방법
    - 결과적으로 미분 값이 작은 곳(극솟값, local minimum)으로 수렴하는 특징이 있음
    - 정확한 파라미터가 아닌 최적 파라미터의 근사치를 찾는 기법

$$w_{new} = w_{old} - \eta\cdot\frac{∂Loss}{∂w_{old}}$$

- 기울기는 손실 함수의 각 파라미터에 대한 편미분 값으로, 특정 파라미터가 손실 함수에 얼마나 영향을 미치는지를 나타냄
    - 기울기가 양수(녹색)라면 가중치의 값이 커질 수록 함수의 값이 커진다는 것을 의미
    - 기울기가 음수(적색)라면 가중치의 값이 커질 수록 함수의 값이 작아진다는 것을 의미

#### 경사 하강법 유형
- 배치 경사 하강법 (Batch Gradient Descent)
    - 배치 경사 하강법은 전체 훈련 데이터 셋을 사용하여 한 번의 기울기를 계산하고이를 통해 파라미터를 에폭 당 1회 업데이트
    - 안정적이며, 정확한 기울기를 계산하지만 매우 큰 데이터세트의 경우 계산 비용이 높아질 수 있으며, 메모리 요구 사항이 증가됨
- 확률적 경사 하강법 (Stochastic Gradient Descent, SGD)
    - 확률적 경사 하강법은 매 반복마다 한 개의 데이터 포인트를 사용하여 기울기를 계산하고, 파라미터를 에폭 당 N회 업데이트
    - 계산이 빠르고, 적은 메모리로도 수행 가능하며, 자주 업데이트하므로 최적화 과정에서 지역 최솟값을 탈출할 가능성이 높지만 기울기의 변동이 크기 때문에 손실 함수의 수렴이 불안정함
- 미니배치 경사 하강법 (Mini-batch Gradient Descent)
    - 미니배치 경사 하강법은 훈련 데이터세트를 작은 배치로 나누어 각 배치에 대해 기울기를 계산하고 파라미터를 에폭 당 N/B회 업데이트
    - 배치 경사 하강법과 확률적 경사 하강법의 절충안. 계산 효율성과 안정성을 모두 제공하지만 배치 크기 설정에 따라 성능이 달라질 수 있음
- 모멘텀을 추가한 경사 하강법
    - 기울기의 누적치를 사용하여 진동을 줄이고, 더 빠르게 최적점에 도달하도록 돕는 방법
    - Momentum, Nesterov Momentum을 적용한 옵티마이저들이 이 방식에 해당
    - 옵티마이저의 `momentum` 인자를 설정하여 사용
- 적응형 학습률을 사용하는 경사 하강법
    - 각 매개변수에 대해 학습률을 동적으로 조정하는 방식으로 경사 하강법의 단점을 보완하며, 데이터의 스케일에 맞춰 학습률을 조정하여 더 빠르고 효율적인 학습을 가능하게 함
    - AdaGrad, RMSprop, Adam과 같은 옵티마이저들이 이 방식에 해당

### 역전파
- 역전파(Backpropagation)는 인공신경망에서 학습을 수행할 때 사용되는 핵심 알고리즘
    - 역전파는 순전파 과정에서 계산된 손실 값을 거꾸로 전파하여 각 가중치가 손실에 얼마나 기여했는지 계산하는 과정
- 역전파는 모델이 예측한 결과와 실제 결과 간의 차이를 기반으로 각 층에서 손실 함수의 미분 값을 계산
    - 미분 값은 손실 함수의 변화율을 나타내며 특정 가중치의 변화가 손실 함수에 어떤 영향을 미치는지 측정
    - 가중치를 조정하여 손실을 줄이는 방향으로 학습을 진행 (경사 하강법)
- 역전파 과정에서 기울기를 계산하기 위해 체인 룰(Chain Rule)을 사용
    - 각 층의 출력에 대해 연쇄적으로 미분(합성 함수의 미분)을 계산
- 경사 하강법(Gradient Descent)과 같은 최적화 알고리즘을 사용하여, 계산된 미분 값을 이용하여 각 가중치와 편향을 업데이트
    - 이 과정은 모든 가중치와 편향에 대해 적용되며, 모델의 파라미터가 손실을 줄이는 방향으로 조정
    - 학습률은 매 업데이트마다 계산된 미분 값을 각 파라미터에 얼마나 크게 반영할지를 조절
    - 너무 크면 최적화 과정이 불안정해지고, 너무 작으면 학습이 너무 느려질 수 있음

### 반복 학습
- 반복 학습(Iteration)은 심층 신경망이 데이터에서 점진적으로 학습하는 과정
- 모델의 성능을 점차 향상시키는 데 중요한 역할을 하며 에포크(epoch)와 배치(batch)와 연관됨
- 에포크는 학습 데이터셋 전체를 모델에 한 번 모두 통과시키는 과정을 의미
    - 모든 학습 데이터를 한 번 사용하여 모델의 파라미터(가중치와 편향)를 업데이트하는 과정
    - 초기에는 모델이 데이터를 제대로 이해하지 못하고 큰 손실을 가지지만, 에포크가 진행될수록 손실이 감소하며, 모델의 예측력이 향상
- 너무 많은 에포크를 사용하면 모델이 학습 데이터에 과적합(overfitting)될 수 있음
    - 과적합은 모델이 학습 데이터에 너무 특화되어, 새로운 데이터에 대한 일반화 성능이 떨어지는 현상
    - 너무 많은 에포크로 인한 과적합을 방지하기 위해 조기 종료 기법을 사용할 수 있음
    ```
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=0, restore_best_weights=False)
    ```
    - `monitor` : 모니터링할 값
    - `patience` : 개선되지 않는 epoch수
    - `restore_best_weights` : 최상의 가중치를 복원
    - fit()메서드에서 `callback`인자로 지정
- 배치는 데이터셋의 일부를 의미하며, 모델은 한 번에 이 배치 크기만큼의 데이터를 사용해 가중치를 업데이트
    - 배치 크기가 전체 데이터셋 크기와 동일한 경우, 한 번의 에포크에서 데이터셋 전체를 사용해 한 번의 가중치 업데이트를 수행
    - 배치 크기가 32인 경우 전체 데이터를 32개씩 나누어 한 번의 에포크에서 각각의 배치마다 가중치 업데이트를 수행
- 배치 크기를 통해 학습의 효율성을 조절할 수 있음
    - 작은 배치는 학습 속도를 높이지만 기울기의 변동성이 커질 수 있음
    - 큰 배치는 학습이 안정적이지만 시간이 더 오래 걸릴 수 있음
