# 텍스트 데이터
- 텍스트 데이터는 사람들이 사용하는 자연어로 작성된 글, 문장, 단어 등의 정보를 포함하는 **비정형 데이터**
- 텍스트 데이터는 구조화된 숫자나 표 형식의 데이터와 달리, **언어로 표현된 의미와 문맥을 포함**하고 있음
- 텍스트 데이터의 주요 특징
    - 고정된 형식이 없으며, 다양한 길이와 형태로 존재
    - 디지털화된 문서뿐만 아니라, 종이 문서에서 스캔 된 텍스트, 오디오에서 텍스트로 변환된 대화 등도 포함
    - 사람이 사용하는 자연어로 구성되어 있어 이를 분석하기 위해서는 자연어 처리(NLP: Natural Language Processing) 기술이 필요
    - 단어와 문장이 결합되어 **특정한 의미를 전달하기 때문에문맥에 따라 단어의 의미가 달라질 수 있으며, 문장의 구조나 단어의 순서도 중요한 정보**를 담고 있음

### 텍스트 데이터 수집
- 텍스트 데이터 수집은 필요한 데이터를 다양한 출처에서 확보하는 단계로 **웹 스크래핑(Web Scraping), 데이터베이스(Database), API 활용(API Integration), 기타 소스 확보 방법** 등의 방법을 사용할 수 있음
    - 웹 스크래핑(Web Scraping): 웹 페이지에서 텍스트 데이터를 자동으로 수집하는 방법으로 Python의 BeautifulSoup, Scrapy, Selenium 같은 라이브러리를 사용할 수 있음
    - 데이터베이스(Database): 이미 존재하는 데이터베이스에서 텍스트 데이터를 쿼리하여 수집하는 방법
    - API 활용(API Integration): 특정 웹 서비스가 제공하는 API를 사용하여 데이터를 수집하는 방법
    - 기타 소스 확보 방법: 이메일, 설문 조사 응답, 문서 스캔 등을 통해 텍스트 데이터를 수집하는 방법으로 OCR(Optical Character Recognition)을 사용하면 종이 문서에서 텍스트를 디지털 데이터로 변환할 수 있음

### 텍스트 데이터 정제
- 수집된 텍스트 데이터는 종종 **노이즈**(불필요한 데이터)를 포함하고 있으며, 이를 **정제하는 과정이 필요**
- 데이터 정제는 분석의 정확도를 높이기 위한 중요한 단계로 데이터 클렌징(Data Cleansing)과 데이터 필터링(Data Filtering)이 대표적임
    - **데이터 클렌징**(Data Cleansing): 텍스트 데이터에서 불필요한 부분(HTML 태그, 특수 문자, 불필요한 공백 등)을 제거하는 과정으로 오탈자 교정, 반복되는 단어 제거, 특수 기호 제거 등의 작업이 포함되어 있음
    - **데이터 필터링**(Data Filtering): 특정 기준에 따라 데이터를 필터링하여 분석에 필요한 데이터만 남기는 작업으로 특정 언어로 작성된 텍스트만 필터링하거나, 특정 주제에 관련된 텍스트만 추출하는 방법 등의 작업이 포함되어 있음

### 텍스트 데이터 전처리
- 데이터 전처리 (Data Preprocessing)는 정제된 텍스트 데이터를 분석할 수 있도록 가공하는 과정으로 텍스트 데이터가 머신러닝 모델이나 기타 분석 도구에서 효과적으로 활용될 수 있도록 준비하는 과정
    - **토큰화**(Tokenization): 텍스트를 분석하기 쉬운 작은 단위로 분할하는 과정으로 문장을 단어 단위로 분할하는 작업
    - **정규화**(Normalization): 텍스트 데이터를 일관된 형태로 변환하는 과정으로 대소문자 통일, 불필요한 구두점 제거, 동의어 처리, 어간 추출(Stemming), 표제어 추출(Lemmatization) 등의 작업
    - **데이터 증강**(Data Augmentation): 텍스트 데이터의 양과 다양성을 늘리기 위해 기존 데이터를 변형하거나 새로운 데이터를 생성하는 방법으로 문장의 단어 순서 변경, 동의어를 사용하여 같은 의미의 다른 문장을 생성하는 방식 등이 존재

## 한국어의 특성
- 한국어는 다른 언어와는 구별되는 몇 가지 독특한 특성을 가지고 있어 한국어 텍스트를 처리할 때 특별한 기술적 접근이 필요
- 한국어는 교착어(agglutinative language)로, 단어에 다양한 접사(어미, 조사 등)가 결합되어 새로운 의미를 만들 수 있음
    - 동사 "가다"에 "갔다", "가는 중이다", "가고 싶다"와 같은 형태가 붙어 여러 의미를 표현
- 한국어는 주어-목적어-서술어(SOV) 구조를 기본으로 하지만, 비교적 어순이 자유로워 문맥에 따라 단어의 순서가 달라지므로, 어순을 고정하는 언어들과 비교했을 때 구문 분석(syntax analysis)이 더 어려움
    - "나는 밥을 먹었다"와 "밥을 나는 먹었다"가 모두 의미 전달이 가능
    - 한국어는 단어 간 띄어쓰기가 엄격하지 않으며, 규칙적이지 않은 경우도 많아 문장을 처리할 때 어디서 단어를 분리해야 할지 명확하지 않을 수 있으며 토큰화 과정이 매우 중요함
    - "한국어자연어처리"라는 문장은 "한국어 자연어 처리" 또는 "한 국어 자연어 처리"로 분리될 수 있음
- 한국어는 존칭과 높임말 체계가 있어 같은 의미의 문장도 상대방에 따라 다른 표현이 필요할 수 있으며, 이 때문에 문장 간의 차이를 고려해야 함
    - "밥 먹었어?"와 "식사하셨습니까?"는 같은 의미이지만, 서로 다른 사회적 맥락에서 사용
- 한국어에서는 조사와 접미사가 문장의 의미를 결정하는 데 중요한 역할을 하며 단어가 주어인지 목적어인지에 따라 완전히 다른 의미를 가지게 됨
    - "사과를 먹다"와 "사과가 먹다"는 "사과"라는 단어가 주어인지 목적어인지에 따라 완전히 다른 의미
- 한국어는 동음이의어나 의미적 중의성이 많아, 문맥을 이해하지 않으면 정확한 의미를 해석하기 어려우며 이를 해결하기 위해서는 단어의 의미를 문맥에 맞게 해석하는 기술이 필요
    - "배"라는 단어는 문맥에 따라 "과일 배", "신체 부위 배", "배(船)"라는 서로 다른 의미로 해석

## 토큰화(Tokenization)
- **토큰화는 문장을 토큰 단위로 분할하는 과정**으로 토큰화 알고리즘에 따라 방법이 상이할 수 있음
    - **최근의 모델은 대부분 형태소 단위로 분리**
- 각 **언어에 최적화된 Tokenizer를 사용**하는 것이 단어를 분할하는데 용이
- Hugging Face의 transformers 라이브러리에서 제공하는 AutoTokenizer는 사용자가 특정 모델의 이름만 제공하면, 그 모델에 적합한 토크나이저를 알아서 불러오는 장점이 있음
- 토크나이저(Tokenization)는 텍스트를 모델이 이해할 수 있는 숫자 시퀀스로 변환하는 과정으로 AutoTokenizer는 이를 자동으로 처리해 주며, 단어를 토큰 단위로 분할하거나 서브워드(subword) 단위로 분할함

토큰화 하는 방법
1. keras에서 제공하는 preprocessing.text.Tokenizer 사용해 워드 인덱스로 바뀜
2. hugging face 에서 이미 만들어진 사전학습된 모델사용(이미 의미를 기반해 분할해 놓은 단어 사전)

## Word2Vec
- 텍스트의 단어들을 주변 단어(context word)와 가운데 단어(target word)로 나누어 단어들간의 의미적 관계를 학습하여 **단어를 벡터화**하는 모델
    - CBOW(Continuous Bag of Word) 방식 : **주변 단어를 이용**하여 가운데 단어를 예측
    - Skip-gram 방식 : **가운데 단어에서 주변 단어**를 예측
    - **유사한 의미를 가진 단어**는 **가깝게 배치**됨
        - 예 : 킹과 퀸은 성별을 제외하고 서로 주변단어로 사용되고, 왕관, 통치, 왕국 등과 같은 단어들이 동일하게 주변단어로 사용되어 킹과 퀸은 가까운 위치로 배치됨
    
    image.png
- CBOW <br>
$ P(\textbf{brown} | the, quick, for, jumps)$
image.png

- Skip-Gram <br>
$ P(the, quick, for, jumps | \textbf{brown})=P(the|\textbf{brown}) \ast P(quick|\textbf{brown}) \ast P(for|\textbf{brown}) \ast P(jums|\textbf{brown})$
image.png

- 단어, 아이템, 또는 카테고리와 같은 이산적인 입력 데이터를 밀집 벡터(dense vector)로 변환해주는 레이어로 주로 자연어 처리에서 사용되며, 단어를 고차원 벡터 공간으로 매핑하는 데 활용
    - 원-핫 인코딩처럼 고차원의 희소 벡터 대신 저차원의 밀집 벡터로 변환해 메모리와 계산 효율성을 높임
- 하나의 큰 행렬(임베딩 테이블)을 유지하며, 각 행은 각 이산적인 입력의 임베딩 벡터를 나타냄
    - 인덱스 값(단어의 고유 번호와 같은 정수)을 입력으로 받으며 해당 인덱스에 해당하는 임베딩 벡터를 반환함
    - 학습을 통해 연관 관계가 높은 항목들이 임베딩 벡터 공간에서 가깝게 배치되도록 함

```
tf.keras.layers.Embedding(input_dim, output_dim, mask_zero=False)
```
- `input_dim` : 데이터 입력의 크기 (vocabulary 크기)
- `output_dim` : Dense의 크기
- `mask_zero` : 입력 데이터의 패딩토큰(0)이 포함되어 있는지 여부

###[예제] 문장의 긍정/부정 분류

````python
texts = [
    'very good nice quality',
    'stop lying',
    'ugly terrible',
    'excellent work',
    'adorable lovely',
    'bad',
    'great nice'
]
y = tf.constant([1, 0, 0, 1, 1, 0, 1])
y = tf.reshape(y, (-1,))
````

# 순환신경망(RNN, Recurrent Neural Network)
- RNN은 순차적 데이터를 처리하고 학습하는 데 특화된 신경망
- 입력 데이터를 처리할 때 이전 상태를 기억하고, 그 정보를 바탕으로 현재 상태를 업데이트하는 방식으로 동작
- RNN은 시간 순서에 따라 데이터가 입력되기 때문에 이전 입력에서 학습한 내용을 현재의 학습에 반영할 수 있는 구조
    - 기존의 신경망은 은닉층의 출력이 출력층 방향으로만 전달되지만 순환 신경망은 출력층 방향으로 전달되는 동시에 은닉층 자신에게 전달되는 구조
- 순환 신경망 구조의 기본 단위는 셀(Cell)로 기존 신경망의 퍼셉트론과 비슷한 역할을 하지만 은닉 상태(hidden state)가 순환하며 연결되는 경로인 순환 경로(Recurrent Path)를 가지고 있음
    - 이전 정보를 기억하기 때문에 Memory Cell이라고 부르기도 함

## Vanilla RNN
```
tf.keras.layers.SimpleRNN(units, activation='tanh')
```
- 순환신경망의 가장 기본적인 형태
- 출력 결과가 직전 계산 결과에 크게 의존함
- 비교적 짧은 시퀀스(sequence)에 대해서만 효과가 있으며 시퀀스 길이가 길어질수록 오래된 정보가 뒤로 충분히 전달되지 못함
    - 장기 의존성 문제(the problem of Long-Term Dependencies) 발생

image.png

## LSTM(Long Short-Term Memory)
```
tf.keras.layers.LSTM(units, activation='tanh', recurrent_activation='sigmoid')
```
- LSTM은 긴 시퀀스 데이터에서 발생하는 장기 의존성(Long-term Dependencies) 문제와 기울기 소실 문제를 해결하기 위해 게이트(Gate)라는 개념을 도입한 구조
- LSTM은 네트워크의 중요한 정보를 오랜 시간 동안 기억하고, 필요에 따라 잊어버릴 수 있도록 세 가지 주요 게이트로 구성
    - 입력 게이트(Input Gate) : 현재 시점의 입력 정보가 얼마나 새롭게 기억될지를 결정
    - 망각 게이트(Forget Gate) : 이전 시점에서 전달된 기억 중에서 얼마나 잊어버릴지를 결정
    - 출력 게이트(Output Gate) : 현재 시점의 기억이 얼마나 출력으로 반영될지를 결정
- LSTM은 은닉 상태 외에도 셀 상태(Cell State)라는 추가적인 상태를 유지하여, 장기 정보를 기억

image.png

## GRU(Gated Recurrent Unit)
```
tf.keras.layers.GRU(units, activation='tanh', recurrent_activation='sigmoid')
```
- GRU는 LSTM보다 더 간결한 구조를 가지면서도 유사한 성능을 제공함
- GRU는 LSTM과 비교했을 때, 게이트의 수가 적고 셀 상태 없이 은닉 상태만을 사용함
- GRU는 업데이트 게이트(Update Gate)와 리셋 게이트(Reset Gate)라는 두 가지 게이트를 사용하여 정보의 흐름을 조절
    - 업데이트 게이트 (Update Gate)
        - 현재 시점의 은닉 상태를 얼마나 업데이트할지를 결정함
        - LSTM의 입력 게이트와 망각 게이트를 하나로 합친 역할
    - 리셋 게이트 (Reset Gate)
        - 이전 은닉 상태를 얼마나 리셋(무시)할지를 결정함
        - 현재 시점에서 새로운 입력을 얼마나 반영할지를 조절

## 순환신경망 역전파 과정
- 순환신경망은 시간축을 따라 역전파가 이루어지기 때문에 기존의 역전파 알고리즘과 유사한 BPTT(Backpropagation Through Time) 방법을 사용함
    - 네트워크의 은닉 상태가 시간에 따라 반복적으로 업데이트 되기 때문에 각 시점에서의 출력은 이전 시점의 은닉 상태에 의존하게 됨
    - 시간 의존성을 고려하여, 출력의 손실을 네트워크 전체에 걸쳐 역전파하는 과정이 필요
- 시간축을 따라 네트워크를 펼친 뒤, 각 타임 스텝의 출력과 손실을 이용해 기울기를 계산하는 방식으로 손실은 각 타임 스텝에서 계산되고, 마지막 출력뿐만 아니라 모든 출력에서 역방향으로 전파되어 네트워크를 학습시킴
    - 모든 타임 스텝에서 동일한 가중치를 공유하기 때문에, 역전파 과정에서는 각 타임 스텝에서 계산된 기울기가 시간 축을 따라 누적되어 파라미터의 기울기를 합산
    - 합산된 기울기는 한 번에 업데이트되며, 동일한 가중치가 모든 타임 스텝에 다시 적용
- $W_y, W_h, W_x$의 기울기 <br>
    -
$\frac{\partial{Loss}}{\partial{W_y}} = \frac{\partial{Loss}}{\partial{\hat{y}}}\frac{\partial{\hat{y}}}{\partial{W_y}} = (\hat{y}-y)h_t$ <br>
    -
$\frac{\partial{Loss}}{\partial{W_h}} = \sum_{t=1}^T\frac{\partial{Loss}}{\partial{h_t}}\frac{\partial{h_t}}{\partial{h_{t-1}}}\frac{\partial{h_{t-1}}}{\partial{W_h}}$<br>
    -
$\frac{\partial{Loss}}{\partial{W_x}} = \sum_{t=1}^T\frac{\partial{Loss}}{\partial{h_t}}\frac{\partial{h_t}}{\partial{W_x}}$

###[예제] LSTM을 이용한 단어 글자 예측(One to One)

````python
# 학습할 단어 목록
words = ['actor', 'angel', 'april', 'bacon', 'birth', 'black', 'cacao', 'child', 'count', 'drink']


# 글자 사전 생성 함수
def build_vocab(data):
    tokens = []
    for word in data:
        tokens.extend(list(word))
    # 중복 제거 후 글자 사전 생성
    return {k:v for v, k in enumerate(set(tokens))}

vocab = build_vocab(words)
vocab
````

````python
# idx2word
idx2word = {i:w for w, i in vocab.items()}
idx2word
````

````python
# 단어를 글자로 토큰화
seqs = [[vocab[c] for c in word] for word in words]
seqs
````

````python
# seqs(토큰화된 단어 데이터)를 원핫 벡터화
from tensorflow import one_hot

tf_seqs = tf.constant(seqs)

x = one_hot(tf_seqs[:,:-1], depth=len(vocab))
y = tf_seqs[:, -1]
x[:1], x.shape, y.shape
````

````python
# 모델 구성
class NextChar(tf.keras.Model):
  def __init__(self):
    super().__init__()
    self.lstm = layers.LSTM(32)
    self.fc = layers.Dense(len(vocab), activation='softmax')

  def call(self, x):
    out = self.lstm(x)
    out = self.fc(out)
    return out
````

````python
# 모델 학습
lstm = NextChar()
lstm.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
lstm.fit(x, y, epochs=100, verbose=0)
````

````python
# 모델 예측
y_prob = lstm.predict(x)
y_pred = tf.argmax(y_prob, axis=1)
y_pred
````

````python
# 예측 결과 출력
print(f'원본 데이터: {words}')
print(f'입력 데이터: {["".join([idx2word[i] for i in seq[:-1]]) for seq in seqs]}')
print(f'예측한 글자: {[idx2word[i] for i in y_pred.numpy()]}')
````
