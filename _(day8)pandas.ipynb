```python
# Indexer
# : - Pandas에서 데이터 선택은 기본적인 인덱싱과 슬라이싱보다는 Indexer를 이용하는 방법 권장
#     why? 기본 인덱싱, 슬라이싱으로 선택된 데이터는 값을 추가/수정 시 오류 발생 가능


#   - Pandas Indexer : loc, iloc, at, iat

#     * loc, at      :   이름을 기반으로 인덱싱
#     * iloc, iat    :   위치를 기반으로 인데싱 (i는 integer)
#     * at, iat      :   한개의 값만 반환

# < 문법 >
# dataframe.인덱서[행인덱스, 열인덱스]
```

```python
# 기본 인덱싱과 슬라이싱 방법으로 선택된 데이터는 원본 데이터와 연결되지 않거나, 참조가 아닌 복사본을 반환하는 경우가 많습니다. 이로 인해 데이터를 수정하려고 할 때 예기치 않은 오류가 발생할 수 있습니다. 
# 데이터를 슬라이싱하여 subset에 할당했을 때, subset은 df['Age']의 복사본이 됩니다. 
# 이 복사본에 대해 값을 변경하려고 할 때 "SettingWithCopyWarning" 경고가 발생하고, 실제 원본 데이터는 수정되지 않을 수 있습니다. 
# 인덱서(indexer)를 사용하면 데이터가 참조되기 때문에 원본 데이터에 대해 직접 작업할 수 있습니다. 
# 인덱서를 사용하면 값 수정, 추가, 삭제 등의 작업이 안전하고 정확하게 수행합니다.
```
