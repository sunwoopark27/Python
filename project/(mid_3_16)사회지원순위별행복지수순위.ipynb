```python
import pandas as pd
```

```python
df = pd.read_csv('/content/drive/MyDrive/WHR_2018_2022.csv')
df.head()
```

```python
# 행복지수 순위 컬럼 추가
df['행복지수 순위'] = df['happiness_score'].rank(method='dense', ascending=False).astype(int)

# social_support 기준 상위 10개 국
top10_social_support = df.sort_values(by='social_support', ascending=False).head(10).reset_index(drop=True)

# 상위 10개국의 행복지수 순위 출력
print(top10_social_support[['country', 'social_support', 'happiness_score', '행복지수 순위']])
```

```python
# social_support 기준 하위 10개 국
bottom10_social_support = df.sort_values(by='social_support', ascending=True).head(10).reset_index(drop=True)

# 하위 10개국의 행복지수 순위 출력
print(bottom10_social_support[['country', 'social_support', 'happiness_score', '행복지수 순위']])
```
