### 2018-2024 까지 데이터 병합

**모든 년도를 concat** 해서 어떤 변수가 변함으로써 변화가 이루어지는지 봄.

```python
# import numpy as np

# # 'df_2023'과 'df_2024'에서 'Logged GDP per capita' 값을 원래 값으로 변환
# df_2023['gdp_per_capita'] = np.exp(df_2023['Logged GDP per capita'])
# df_2024['gdp_per_capita'] = np.exp(df_2024['Logged GDP per capita'])
```

```python
import numpy as np

# 'gdp_per_capita'를 로그 처리하여 'Logged GDP per capita' 추가
df['Logged GDP per capita'] = np.log(df['gdp_per_capita'])

# 'df_2024'에서 'Logged GDP per capita'를 원래 값으로 변환하고 'gdp_per_capita' 컬럼을 추가
df_2024['gdp_per_capita'] = np.exp(df_2024['Logged GDP per capita'])


# 컬럼 이름을 맞추기 위해 df_2023, df_2024 의 컬럼명을 수정
df_2023 = df_2023.rename(columns={
    'Country name': 'country',
    'Ladder score': 'happiness_score'
})

df_2024 = df_2024.rename(columns={
    'Country name': 'country',
    'Ladder score': 'happiness_score',
    'Social support': 'social_support',
    'Healthy life expectancy': 'healthy_life_expectancy',
    'Freedom to make life choices': 'freedom_to_make_life_choices',
    'Generosity': 'generosity',
    'Perceptions of corruption': 'perceptions_of_corruption',
    'Ladder score in Dystopia': 'dystopia_residual',
})

df_2023 = df_2023.drop(columns=['upperwhisker','lowerwhisker'])
df_2024 = df_2024.drop(columns=['Standard error of ladder score',	'upperwhisker',	'lowerwhisker'])
```

```python
# 병합
df_1824 = pd.concat([df, df_2023, df_2024], axis=0, ignore_index=True, join='outer')

# 결과 확인
df_1824.head()
```

2023년에 변수별 기여도를 측정하는 방식이 바뀌어 정확히 보기 어려움

#### 연도별 상관분석

```python
# 연도별로 상관계수를 계산하려면, 수치형 데이터만 선택
years = df_1824['year'].unique()  # 연도 리스트
for year in years:
    df_year = df_1824[df_1824['year'] == year]

    # 수치형 데이터만 선택
    df_year_numeric = df_year.select_dtypes(include=['float64', 'int64'])

    # 상관계수 계산
    correlation = df_year_numeric.corr()

    print(f"{year}년 상관분석:")
    print(correlation['happiness_score'])  # happiness_score와의 상관계수만 출력
    print("\n")
```

```python
df_korea_1824 = df_1824[df_1824["country"] == "South Korea"]
df_korea_1824 = df_korea_1824[df_korea_1824["year"].isin([2018, 2019, 2020, 2021, 2022, 2024])]
```

```python
# 변수와 제목 리스트
variables = ["happiness_score", "gdp_per_capita", "social_support",
             "healthy_life_expectancy", "freedom_to_make_life_choices",
             "generosity"]
titles = ["Happiness Score", "GDP per Capita", "Social Support",
          "Healthy Life Expectancy", "Freedom to Make Life Choices",
          "Generosity"]

# 3x2 subplot 설정
fig, axes = plt.subplots(3, 2, figsize=(14, 12))

# 각 변수에 대해 시각화
for idx, var in enumerate(variables):
    row = idx // 2  # subplot의 행 번호
    col = idx % 2   # subplot의 열 번호
    ax = axes[row, col]  # 해당 위치의 Axes 객체를 선택

    # 해당 변수의 값 시계열로 그리기
    ax.plot(df_korea_1824["year"], df_korea_1824[var], marker='o', color='b', linestyle='-', label=var)

    # 제목, 레이블 설정
    ax.set_title(titles[idx], fontsize=12)
    ax.set_xlabel("Year", fontsize=10)
    ax.set_ylabel(var, fontsize=10)
    ax.grid(True)

# 그래프 레이아웃 최적화
plt.tight_layout()
plt.show()

# happiness score은 무시
```
