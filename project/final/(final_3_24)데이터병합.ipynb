## 4개의 데이터 병합

    - Chicago_Crimes_2012_to_2017.csv
    - Chicago_Crimes_2008_to_2011.csv
    - Chicago_Crimes_2005_to_2007.csv
    - Chicago_Crimes_2001_to_2004.csv

```python
# 데이터가 모두 같은 컬럼을 가짐
df = pd.concat([one, two, three, four], axis=0, ignore_index=True)
```

```python
df.head()
```

## Date 컬럼


    - Date 컬럼
      date/time 형식으로 되어있음
    - date와 time 컬럼 추가해서 따로 분석할 수 있게 나누기

```python
df.dtypes
```

```python
# Date 컬럼의 형식 object(문자열) -> 형식 변환
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y %I:%M:%S %p')
```

```python
# 날짜(Date)와 시간(Time) 나누기
df['F_Date'] = df['Date'].dt.strftime('%Y-%m-%d')
df['Time'] = df['Date'].dt.strftime('%H:%M:%S')

# 결과 출력
print(df[['F_Date', 'Time']])
```

```python
df.to_csv('Chicago_Crimes_2001_to_2017.csv', index=False)
```
