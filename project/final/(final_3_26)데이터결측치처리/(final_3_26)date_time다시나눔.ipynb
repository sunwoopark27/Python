## Date 컬럼


    - Date 컬럼
      date/time 형식으로 되어있음
    - date와 time 컬럼 추가해서 따로 분석할 수 있게 나누기

```python
chicago.dtypes
```

```python
# Date 컬럼의 형식 object(문자열) -> 형식 변환
chicago['Date'] = pd.to_datetime(chicago['Date'], format='%m/%d/%Y %I:%M:%S %p')
```

```python
# 날짜(Date)와 시간(Time) 나누기
chicago['F_Date'] = chicago['Date'].dt.strftime('%Y-%m-%d')
chicago['Time'] = chicago['Date'].dt.strftime('%H:%M:%S')

# 결과 출력
print(chicago[['F_Date', 'Time']])
```
