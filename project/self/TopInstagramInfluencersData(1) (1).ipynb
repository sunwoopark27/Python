```python
import pandas as pd
insta = pd.read_csv('/content/top_insta_influencers_data.csv')
insta.head()
```

### 팔로워 TOP 10

```python
#팔로워 Top 10
insta.sort_values(by='followers',ascending=False).head(10)
```

### 포스트당 평균 참여율 컬럼 생성

```python
# 문자열을 숫자형으로 변환 (변환 불가능한 값은 NaN으로 처리)
insta['avg_likes'] = pd.to_numeric(insta['avg_likes'], errors='coerce')
insta['followers'] = pd.to_numeric(insta['followers'], errors='coerce')

# 참여율 계산
insta['engagement_rate'] = insta['avg_likes'] / insta['followers']
```
