```python
# 1로 만들기
# 규칙이 나와있음
# X 가 3으로 나누어 떨어지면 3으로 나눈다.
# X rk 2로 나누어 떨어지면 2로 나눈다.
# dp[n] : 1에서 n을 만들 때 연산의 최소 횟수

# 점화식 바탕으로 코드 작성
N = int(input())

# [1] dp[n] : 1에서 n을 만들때 사용되는 연산의 최소회수
# [2] dp테이블 생성 및 초기화
dp = [0]*(N+1)
dp[1] = 0      # 1부터 시작하므로 이곳은 연산회수 0

for n in range(2, N+1) :
    dp[n] = dp[n-1] + 1
    if n%2 == 0:
      dp[n] = min(dp[n],dp[n//2]+1)
    if n%3 == 0:
      dp[n] = min(dp[n],dp[n//3]+1)

print(dp[N])
```
